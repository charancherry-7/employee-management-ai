```sql
USE CATALOG employee_management;

USE SCHEMA fullstack;

CREATE TABLE IF NOT EXISTS employees
(
    employee_id INT,
    first_name STRING,
    last_name STRING,
    email STRING,
    phone STRING,
    department STRING,
    designation STRING,
    salary DOUBLE,
    date_of_joining DATE,
    status STRING,
    created_at TIMESTAMP,
    updated_at TIMESTAMP
);
```
